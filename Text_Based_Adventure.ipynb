```python
def responseValidator(response):
  while response != "Y" and response != "N":
    response = input("Invalid Response: please type 'Y' or 'N': ")
  return response
```

```python
def intro():
  response = input("You wake up, dreading another day of working at Gandalf's. Do you go to work? Y or N: ")
  return response
```

```python
def choice1(response):
  response = responseValidator(response)

  if response == "Y":
    print("You go to work.")
    return(response)

  elif response == "N":
    print("You stay home.")
    return(response)
```

```python
def gandalfWork():
  print("Another day another wizard")
  print("You hear a clank... do you investigate?")
  response = input("Y or N: ")
  return response
```

```python
def choice2(response):
  response = responseValidator(response)

  if response == "Y":
    print("You go check out the noise.")
    return response

  elif response == "N":
    print("You stay put.")
```

```python
def register():
  print("Could anything be more monotonous?")
```

```python
def main():
  response = intro()
  response = choice1(response)
  if response == "Y":
    gandalfWork()
  elif response == "N":
    print("GAME OVER!!!!!")
  response = choice2(response)
  if response == "Y":
    register()

```

```python
main()
```
