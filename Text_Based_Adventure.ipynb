```python
def responseValidator(response):
  while response != "Y" and response != "N":
    response = input("Invalid Response: please type 'Y' or 'N': ")
  return response
```

```python
endings = ["Ending 1: W Tyreculis", "Ending 2: No Sheckels?", "Ending 3: RICH!", "Ending 4: Another day, Another Wizard...", "Ending 5: Should've been a Hero", "Ending 6: A Raise?", "Ending 7: I Hate It Here"]
```

```python
def intro():
  response = input("You wake up, dreading another day of working at Gandalf's. Do you go to work? Y or N: ")
  return response
```

```python
def choice1(response):
  response = responseValidator(response)

  if response == "Y":
    print("You go to work.")
    return(response)

  elif response == "N":
    print("You stay home.")
    return(response)
```

```python
def gandalfWork():
  print("Another day another wizard")
  response = input("You hear a clank... do you investigate? Y or N: ")
  return response
```

```python
def choice2(response):
  response = responseValidator(response)

  if response == "Y":
    print("You go check out the noise.")
    return response

  elif response == "N":
    print("You stay put.")
    return response
```

```python
def aisle():
  print("What could be going on here?")
  print("Upon getting to the aisle, you see two wizards seemingly... robbing the place!")
  response = input("They haven't noticed you yet, you could run away, or you could try to confront them. Do you want to confront them? Y or N: ")
  return response
```

```python
def choice3b(response):
  response = responseValidator(response)

  if response == "Y":
    print("You steel your nerves, and confront the thieving wizards.")
    return(response)

  elif response == "N":
    print("You decide it's not worth it, and slink away.")
    return(response)
```

```python
def combat():
  print("One of the wizards begin chanting a spell. You recognize it as Fireball.")
  response = input("You could try Counter Spell, but your spellcasting has never been that strong. There are also some cans in the aisle too. Do you try to Counter? Y or N: ")
  return response
```

```python
def choice4(response):
  response = responseValidator(response)

  if response == "Y":
    print("You quickly prepare Counter Spell.")
    return(response)

  elif response == "N":
    print("You grab a can from one of the shelves and prepare to throw it.")
    return(response)
```

```python
def register():
  print("Could anything be more monotonous?")
  response = input("Eriboser, a regular, comes up asking for a runic lottery ticket. You know he's a chronic gambler, and you probably shouldn't give him the ticket. Do you? Y or N: ")
  return response
```

```python
def choice3a(response):
  response = responseValidator(response)

  if response == "Y":
    print("You hand over the lottery ticket.")
    return(response)

  elif response == "N":
    print("You refuse and tell him that they are... not in stock, yeah.")
    return(response)
```

```python
import random

def main():
  response = intro()
  response = choice1(response)
  if response == "Y":
    response = gandalfWork()
    response = choice2(response)
    if response == "Y":
      response = aisle()
      response = choice3b(response)
      if response == "Y":
        response = combat()
        response = choice4(response)
        if response == "Y":
          print("You cast your spell just as the Fireball flies but... this doesn't look like Counter Spell? The Fireball bounces back towards the robbers and incinerates them. That was a Reflect Spell!")
          print("The cameras recorded everything, so you're off the hook with the police. The manager was so impressed, he gave you a raise!")
          print(endings[5])
        elif response == "N":
          print("Turns out, your throwing is also terrible. You completely miss, and eat a Fireball to the face. All you can think of is how you just died at a job you hate...")
          print(endings[6])
      elif response == "N":
        print("You snuck outside and called the police. While they were able to stop the robbery, the cameras caught you leaving the scene. You end up fired for inactivity.")
        print(endings[4])


    elif response == "N":
      response = register()
      response = choice3a(response)
      if response == "Y":
        endingB = random.randint(1, 2)
        if endingB == 1:
          print("He... actually won? In fact, he won the jackpot! Eriboser even gives you a share for being his favorite clerk, and now you have a ton of sheckels.")
          print(endings[2])
        elif endingB == 2:
          print("Just like before, he lost. He storms out muttering something about it being rigged, and nothing else really happens. There were a pair of shady wizards leaving, but you thought nothing of it.")
          print(endings[3])

      print("Game Over")



  elif response == "N":
    endingA = random.randint(1, 2)
    if endingA == 1:
      print("Tyreculis was able to pick up your shift since he still owes you for the sheckels you lent him")
      print(endings[0])
    else:
      print("You got fired because you didn't show up. How will you make sheckels now?")
      print(endings[1])

    print("Game Over")

```

```python
main()
```
